#Data Transformation

###Adding Male row to the dataset

```python
import pandas as pd

# Load the dataset
file_path = 'Education Data.xlsx'
sheet_name = 'Institutes,Enrollments & Teache'
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Filter Total and Female rows
total_df = df[df['Gender'] == 'Total']
female_df = df[df['Gender'] == 'Female']

# Merge Total and Female data
merged_df = pd.merge(total_df, female_df, on=['Year', 'Institution Type'], suffixes=('_Total', '_Female'))

# Calculate Male values
# Convert columns to numeric before subtraction
merged_df['InstitutesNumber_Male'] = pd.to_numeric(merged_df['InstitutesNumber_Total'], errors='coerce') - pd.to_numeric(merged_df['InstitutesNumber_Female'], errors='coerce')
merged_df['Enrollments_Male'] = pd.to_numeric(merged_df['Enrollments_Total'], errors='coerce') - pd.to_numeric(merged_df['Enrollments_Female'], errors='coerce')
merged_df['Teachers_Male'] = pd.to_numeric(merged_df['Teachers_Total'], errors='coerce') - pd.to_numeric(merged_df['Teachers_Female'], errors='coerce')

# Create Male rows
male_df = merged_df[['Year', 'Institution Type']].copy()
male_df['Gender'] = 'Male'
male_df['InstitutesNumber'] = merged_df['InstitutesNumber_Male']
male_df['Enrollments'] = merged_df['Enrollments_Male']
male_df['Teachers'] = merged_df['Teachers_Male']

# Combine original data with new Male rows
updated_df = pd.concat([df, male_df], ignore_index=True)

# Save to a new Excel file
output_path = 'Modified_Education_Data.xlsx'
updated_df.to_excel(output_path, sheet_name=sheet_name, index=False)

print(f"Modified file saved at: {output_path}")
```

# **Indicator 4.5.1: Parity indices (female/male, rural/urban, etc.) for all education indicators that can be disaggregated**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Specify the file path
file_path = 'Education Data.xlsx'

InsEnroll_df = pd.read_excel(file_path, sheet_name='Institutes,Enrollments & Teache')
literacy_df = pd.read_excel(file_path, sheet_name='Literacy Rates')

# -----------------------------------------------
# 1. GPI for Enrollment
# -----------------------------------------------

# Pivot table for enrollments by gender
pivot_enrollments = InsEnroll_df.pivot_table(
    index=['Year', 'Institution Type'],
    columns='Gender',
    values='Enrollments',
    aggfunc='sum'
).reset_index()

# Calculate GPI for Enrollment (Female / Male)
pivot_enrollments['GPI'] = pivot_enrollments['Female'] / pivot_enrollments['Male']

# Plot GPI for Enrollment
plt.figure(figsize=(12, 6))
barplot = sns.barplot(x='Institution Type', y='GPI', data=pivot_enrollments, palette='coolwarm')
plt.title('Gender Parity Index (GPI) for Enrollment across Education Stages')
plt.xlabel('Education Stage')
plt.ylabel('GPI (Female / Male)')
plt.xticks(rotation=45, ha='right')
plt.axhline(1.0, color='grey', linestyle='--')  # Reference line for parity

# Add data labels
for p in barplot.patches:
    barplot.annotate(f'{p.get_height():.2f}',
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='bottom', fontsize=10, color='black')

plt.show()

# -----------------------------------------------
# 2. GPI for Literacy Rates
# -----------------------------------------------

# Calculate GPI for Literacy Rates (Female / Male)
literacy_df['GPI'] = literacy_df['Female'] / literacy_df['Male']

# Handle division by zero
literacy_df['GPI'] = literacy_df['GPI'].replace([float('inf'), -float('inf')], 0)

# -----------------------------------------------
# 3. GPI Analysis: Urban vs Rural
# -----------------------------------------------

# Filter Urban and Rural data
urban_rural_df = literacy_df[literacy_df['Province/Area'].str.contains(r'\bUrban\b|\bRural\b', regex=True)]

plt.figure(figsize=(12, 6))
barplot_urban_rural = sns.barplot(x='Province/Area', y='GPI', data=urban_rural_df, palette='coolwarm')
plt.title('Gender Parity Index (GPI) for Literacy Rates (Urban vs Rural)')
plt.xlabel('Urban vs Rural')
plt.ylabel('GPI (Female / Male)')
plt.xticks(rotation=45, ha='right')
plt.axhline(1.0, color='grey', linestyle='--')

# Add data labels
for p in barplot_urban_rural.patches:
    barplot_urban_rural.annotate(f'{p.get_height():.2f}',
                                 (p.get_x() + p.get_width() / 2., p.get_height()),
                                 ha='center', va='bottom', fontsize=10, color='black')

plt.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

# Function to create a gauge chart
def create_gauge_chart(value, title, target=90):
    fig = go.Figure(go.Indicator(
        mode="gauge+number",
        value=value,
        domain={'x': [0, 1], 'y': [0, 1]},
        title={'text': title, 'font': {'size': 20}},
        gauge={
            'axis': {'range': [0, 100], 'tickwidth': 1, 'tickcolor': "black"},
            'bar': {'color': "royalblue"},
            'steps': [
                {'range': [0, target * 0.5], 'color': "lightcoral"},
                {'range': [target * 0.5, target * 0.8], 'color': "yellow"},
                {'range': [target * 0.8, 100], 'color': "lightgreen"}
            ],
            'threshold': {
                'line': {'color': "green", 'width': 4},
                'thickness': 0.75,
                'value': target
            }
        }
    ))
    fig.update_layout(font={'color': "black", 'family': "Arial"})
    fig.show()

# Load Facilities Data
file_path = 'Education Data.xlsx'
sheet_name = 'Facility'
Facility_df = pd.read_excel(file_path, sheet_name=sheet_name)

# List of school stages
school_stages = ['Primary', 'Middle', 'High', 'Higher Sec.']

# Filter data for Pakistan (assuming there's a row with "Pakistan" as Region)
overall_facility_data = Facility_df[Facility_df['Region'] == 'Pakistan']

# Loop through each school stage and calculate the aggregated percentage across facilities
for stage in school_stages:
    # Calculate the average percentage for the current stage across all facilities
    stage_access_percentage = overall_facility_data[stage].mean()

    # Create a gauge chart for the current stage
    create_gauge_chart(stage_access_percentage, f"Availability of Basic Facilities: {stage} Schools (Target = 90%)")
```

#**Gender Enrollment Analysis and Chi-Square Test**

```python
import pandas as pd
import plotly.express as px
from scipy.stats import chi2_contingency

# Load the dataset
file_path = 'Education Data.xlsx'
sheet_name = 'Institutes,Enrollments & Teache'
InsEnroll_df = pd.read_excel(file_path, sheet_name=sheet_name)

# Filter data for female and male enrollments
female_data = InsEnroll_df[InsEnroll_df['Gender'] == 'Female']
male_data = InsEnroll_df[InsEnroll_df['Gender'] == 'Male']

# Hypothesis Testing: Test if the proportion of female enrollments significantly differs from males
# Create a contingency table
contingency_table = pd.DataFrame({
    'Female': [female_data['Enrollments'].sum()],
    'Male': [male_data['Enrollments'].sum()]
})

# Perform Chi-Square test of independence
chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Test Results:")
print(f"Chi2: {chi2}, p-value: {p}")

# Visualizing the proportion of enrollments by gender
proportion_data = pd.DataFrame({
    'Gender': ['Female', 'Male'],
    'Enrollment': [
        female_data['Enrollments'].sum(),
        male_data['Enrollments'].sum()
    ]
})

fig_proportion = px.pie(
    proportion_data,
    names='Gender',
    values='Enrollment',
    title='Proportion of Enrollments by Gender',
    template='plotly_white'
)
fig_proportion.show()
```

# Average Transition Ratios by Gender Across Education Stages (2010-2023)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import mplcursors

# Load data
file_path = 'Education Data.xlsx'
InsEnroll_df = pd.read_excel(file_path, sheet_name='Institutes,Enrollments & Teache')

# Ensure consistent formatting for analysis
InsEnroll_df['Year'] = InsEnroll_df['Year'].astype(str)

# Prepare data for Stacked Bar Chart
def prepare_transition_ratios(df):
    # Pivot the data to create separate columns for each Institution Type
    pivot = df.pivot(index=['Year', 'Gender'], columns='Institution Type', values='Enrollments').reset_index()

    # Calculate transition ratios for all stages
    pivot['Primary_to_Middle'] = pivot['Middle Schools'] / pivot['Primary Schools'] * 100
    pivot['Middle_to_High'] = pivot['High Schools'] / pivot['Middle Schools'] * 100
    pivot['High_to_HigherSec'] = pivot['Higher Sec/Inter Colleges'] / pivot['High Schools'] * 100
    pivot['HigherSec_to_DegreeColleges'] = pivot['Degree Colleges'] / pivot['Higher Sec/Inter Colleges'] * 100
    pivot['DegreeColleges_to_Universities'] = (pivot['Universities'] / (pivot['Degree Colleges'] + pivot['Universities'])) * 100

    # Melt the DataFrame for plotting
    transition_columns = [
        'Primary_to_Middle', 'Middle_to_High', 'High_to_HigherSec', 'HigherSec_to_DegreeColleges', 'DegreeColleges_to_Universities'
    ]
    melted_df = pivot.melt(
        id_vars=['Year', 'Gender'],
        value_vars=transition_columns,
        var_name='Transition Stage',
        value_name='Percentage'
    )
    return melted_df

# Prepare the data
transition_data = prepare_transition_ratios(InsEnroll_df)

# Aggregate percentages across all years by stage and gender
aggregated_data = (
    transition_data.groupby(['Transition Stage', 'Gender'])['Percentage']
    .mean()  # Use mean percentage for all years
    .unstack()
    .fillna(0)  # Fill missing values with 0
)

# Plot horizontal stacked bars
fig, ax = plt.subplots(figsize=(14, 8))

# Plot Male bars
bars_male = ax.barh(
    aggregated_data.index,
    aggregated_data['Male'],
    color='#4A90E2',
    label='Male',
    alpha=0.8
)

# Plot Female bars stacked on top of Male
bars_female = ax.barh(
    aggregated_data.index,
    aggregated_data['Female'],
    left=aggregated_data['Male'],  # Stack bars
    color='#F5A623',
    label='Female',
    alpha=0.8
)

# Add labels inside bars for Male
for bar in bars_male:
    width = bar.get_width()
    ax.text(
        width / 2, bar.get_y() + bar.get_height() / 2,
        f'{width:.1f}%',
        ha='center', va='center', color='white', fontsize=10
    )

# Add labels inside bars for Female
for bar, male_width in zip(bars_female, aggregated_data['Male']):
    width = bar.get_width()
    ax.text(
        male_width + width / 2, bar.get_y() + bar.get_height() / 2,
        f'{width:.1f}%',
        ha='center', va='center', color='black', fontsize=10
    )

# Add labels and legend
ax.set_title('Average Transition Ratios by Gender Across Education Stages (2010-2023)', fontsize=16)
ax.set_xlabel('Average Percentage (%)', fontsize=14)
ax.set_ylabel('Transition Stages', fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title='Gender', loc='upper right')
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Load Excel data
file_path = 'Education Data.xlsx'
institutes_data = pd.read_excel(file_path, sheet_name="Institutes,Enrollments & Teache")
expenditures_data = pd.read_excel(file_path, sheet_name="Expernditures")

# Rename and align columns for merging
institutes_data = institutes_data.rename(columns={"Year": "Years"})
expenditures_data["Years"] = expenditures_data["Years"].astype(str)

# Merge datasets on "Years" and "Region"
merged_data = pd.merge(
    institutes_data,
    expenditures_data,
    on="Years",
    how="inner"
)

# Filter relevant columns for analysis
analysis_data = merged_data[[
    "Years",
    "Region",
    "Development Expenditure",
    "Teachers"
]].dropna()

# Convert columns to numeric for regression
analysis_data["Development Expenditure"] = pd.to_numeric(analysis_data["Development Expenditure"], errors='coerce')
analysis_data["Teachers"] = pd.to_numeric(analysis_data["Teachers"], errors='coerce')

# Simple linear regression
X = analysis_data["Development Expenditure"]
y = analysis_data["Teachers"]
X = sm.add_constant(X)  # Add constant for the intercept
model = sm.OLS(y, X).fit()

# Display regression results
print(model.summary())

# Visualization
plt.figure(figsize=(10, 6))
sns.scatterplot(x="Development Expenditure", y="Teachers", data=analysis_data, hue="Region", palette="viridis")
sns.regplot(x="Development Expenditure", y="Teachers", data=analysis_data, scatter=False, color="red")
plt.title("Impact of Development Expenditure on Teacher Availability")
plt.xlabel("Development Expenditure (in Units)")
plt.ylabel("Number of Teachers")
plt.legend(title="Region")
plt.show()
# Convert 'Teachers' and 'Development Expenditure' columns to numeric, forcing errors to NaN
merged_data['Teachers'] = pd.to_numeric(merged_data['Teachers'], errors='coerce')
merged_data['Development Expenditure'] = pd.to_numeric(merged_data['Development Expenditure'], errors='coerce')

# Drop rows with NaN values in relevant columns (if any)
merged_data_clean = merged_data.dropna(subset=['Teachers', 'Development Expenditure'])

# Create a new column for efficiency: Teachers per Development Expenditure
merged_data_clean['Teachers_per_Expenditure'] = merged_data_clean['Teachers'] / merged_data_clean['Development Expenditure']

# Sort data by efficiency
efficient_regions = merged_data_clean.groupby('Region')['Teachers_per_Expenditure'].mean().sort_values(ascending=False)

# Display top regions with high efficiency
print("Regions with the highest Teachers per Unit of Development Expenditure:")
print(efficient_regions)


# Plot efficiency by region
ax = efficient_regions.plot(kind='bar', figsize=(10, 6), color='skyblue')

# Add data labels to the bars
for i in range(len(efficient_regions)):
    ax.text(i, efficient_regions.iloc[i] + 0.05, round(efficient_regions.iloc[i], 2), ha='center', va='bottom')

# Set plot title and labels
plt.title("Regional Efficiency: Teachers per Unit of Development Expenditure")
plt.xlabel("Region")
plt.ylabel("Teachers per Expenditure")

# Show the plot
plt.show()

```
